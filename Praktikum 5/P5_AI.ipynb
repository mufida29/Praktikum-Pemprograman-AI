```python
#deklarasi fungsi
def perkalian(bil1, bil2):
    "Menghitung perkalian 2 bilangan bulat"
    hasil = bil1 * bil2
    return hasil

#program utama
bil1 = int(input('Input Bil 1 : '))
bil2 = int(input('Input Bil 2 : '))

#memanggil fungsi
print('Hasil Perkalian ',bil1,'dengan ',bil2, 'adalah ',perkalian(bil2 = bil1, bil1 = bil2))
```
