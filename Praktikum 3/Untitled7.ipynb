```python

print("=== PROGRAM PENUKARAN NILAI DESIMAL ===")

# a. Input (Menggunakan float untuk bilangan desimal sembarang)
A = float(input("Masukkan nilai desimal A: "))
B = float(input("Masukkan nilai desimal B: "))
C = float(input("Masukkan nilai desimal C: "))
D = float(input("Masukkan nilai desimal D: "))

print("\n[Kondisi Sebelum Ditukar]")
print(f"A = {A}")
print(f"B = {B}")
print(f"C = {C}")
print(f"D = {D}")

A, B, C, D = A, D, B, C

# c. Output
print("\n[Kondisi Setelah Ditukar (A, D, B, C)]")
print(f"Nilai A baru : {A}")
print(f"Nilai B baru : {B}")
print(f"Nilai C baru : {C}")
print(f"Nilai D baru : {D}")
```
